# Start here

This is a short tour of `petrophys`, a Python library for interpreting well logs, the depth-indexed measurements recorded along a borehole, whether by wireline or while drilling (natural radioactivity, bulk density, resistivity, and so on). From those curves the library estimates the things people
actually want: how much clay is in the rock, how much pore space, whether the pores hold water or
hydrocarbons, and how much of the section is worth producing.

The notebooks come in three groups:

* **Foundations** (`01`, `02`): load a well, and run the whole interpretation with one call.
* **One per property** (`03`–`08`): clay volume, porosity, saturation, permeability, lithology,
  net pay. Read the one you need.
* **Deep dives** (`10`–`14`): how the mineral solver, the workflow, the FMI imaging, and the
  core validation actually work under the hood.

Everything runs on the sample wells bundled with the package, so you don't need to find data
first.

```python
import sys, warnings
from pathlib import Path
warnings.filterwarnings("ignore")
REPO = Path.cwd()
while not (REPO / "petrophys").exists() and REPO != REPO.parent:
    REPO = REPO.parent
sys.path.insert(0, str(REPO))
WELL = REPO / "petrophys" / "data" / "geolink" / "30_6-5.las"

import petrophys
from petrophys.io import read_las
print("bundled GeoLink wells:", petrophys.list_samples("geolink"))
well = read_las(WELL)
print("loaded", well.name, "with", well.n_samples, "depth samples and", len(well.curves), "curves")
```

That's the whole setup, a `Well` object with named curves. The next notebook looks at what's inside it; the one after runs a full interpretation.
